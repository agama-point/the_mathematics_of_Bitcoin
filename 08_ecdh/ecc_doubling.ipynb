# Elliptic Curve Point Doubling Example

This notebook demonstrates point doubling on the curve y^2 = x^3 + 7 mod 17.
We start with the point G = (15, 13) and compute 2G step by step.
All formulas are shown in Markdown cells. Python prints only numbers to avoid warnings.

```python
# Prime modulus
p = 17
# Initial point G
x1, y1 = 15, 13
print(f'Initial point G = ({x1}, {y1}) on curve y^2 = x^3 + 7 mod {p}')
```

## Multiplicative Inverse
To divide modulo p, we multiply by the multiplicative inverse.
For n modulo p: n * n^-1 ≡ 1 (mod p).
Using Fermat's theorem: n^-1 ≡ n^(p-2) mod p

```python
def inv(n, p):
    inverse = pow(n, p-2, p)
    print(f'Inverting {n} modulo {p} -> {inverse}')
    return inverse
```

## Step 1: Compute slope
Formula for point doubling:
lambda = (3 * x1^2 + a) / (2 * y1) mod p
Here a = 0, so simplified: lambda = (3 * x1^2) / (2 * y1) mod 17

```python
numerator = 3 * x1**2
denominator = 2 * y1
print(f'Numerator for slope = {numerator}')
print(f'Denominator for slope = {denominator}')
denominator_inv = inv(denominator, p)
slope = (numerator * denominator_inv) % p
print(f'Slope = {slope}')
```

## Step 2: Compute new point 2G
Formulas:
x2 = slope^2 - 2 * x1 mod p
y2 = slope * (x1 - x2) - y1 mod p

```python
x2 = (slope**2 - 2*x1) % p
y2 = (slope * (x1 - x2) - y1) % p
print(f'x2 = {x2}')
print(f'y2 = {y2}')
```

## Step 3: Result
The point doubling gives 2G = (x2, y2)

```python
print(f'Final result: 2G = ({x2}, {y2})')
```
