# Modular Inverse without `pow()`

This notebook explains how to compute the modular multiplicative inverse

\[
n^{-1} \equiv n^{p-2} \pmod p
\]

**without using `pow()`**, by relying on the **Extended Euclidean Algorithm**.

We assume:
- `p` is a prime number
- `gcd(n, p) = 1`

Under these conditions, the inverse always exists.

## Mathematical Background

The modular inverse of `n` modulo `p` is a number `x` such that:

\[
n \cdot x \equiv 1 \pmod p
\]

This is equivalent to the **Diophantine equation**:

\[
n x + p y = 1
\]

If `gcd(n, p) = 1`, such integers `x` and `y` are guaranteed to exist.

The **Extended Euclidean Algorithm** computes exactly these coefficients.

## Extended Euclidean Algorithm

The algorithm finds integers `(x, y)` such that:

\[
a x + b y = \gcd(a, b)
\]

When `gcd(a, b) = 1`, the coefficient `x` is the modular inverse of `a` modulo `b`.

```python
def extended_gcd(a, b):
    if b == 0:
        return a, 1, 0
    g, x1, y1 = extended_gcd(b, a % b)
    return g, y1, x1 - (a // b) * y1
```

### Explanation

- The base case corresponds to:

\[
a \cdot 1 + 0 \cdot 0 = a
\]

- Each recursive step rewrites the equation using the identity:

\[
\gcd(a, b) = \gcd(b, a \bmod b)
\]

- Coefficients are algebraically propagated back up the recursion.

## Modular Inverse Function

Using the extended GCD, we extract the coefficient `x` such that:

\[
n x + p y = 1
\]

Reducing this equation modulo `p` gives:

\[
n x \equiv 1 \pmod p
\]

Therefore, `x mod p` is the modular inverse.

```python
def mod_inverse(n, p):
    g, x, _ = extended_gcd(n, p)
    if g != 1:
        raise ValueError("Inverse does not exist")
    inverse = x % p
    print(f"Inverting {n} modulo {p} -> {inverse}")
    return inverse
```

## Example

Compute the inverse of `3` modulo `11`:

\[
3^{-1} \equiv 4 \pmod{11}
\]
because:

\[
3 \cdot 4 = 12 \equiv 1 \pmod{11}
\]

```python
mod_inverse(3, 11)
```

## Relation to `pow(n, p-2, p)`

- `pow(n, p-2, p)` uses **Fermat's Little Theorem**
- The Extended Euclidean Algorithm is **more general**

Both compute the same value when `p` is prime, but:
- `pow()` is optimized for large primes
- `extended_gcd` works for any modulus where `gcd(n, p) = 1`

Conceptually, the extended GCD exposes the underlying number-theoretic structure.
